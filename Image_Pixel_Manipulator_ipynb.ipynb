```python
filename = next(iter(uploaded))

img = cv2.imread(filename)

if img is None:
    print("Error: Image could not be loaded.")
else:
    print("Image loaded successfully!")
    print("Original image shape:", img.shape)
```

```python
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
height, width, channels = img.shape

print("Height:", height)
print("Width:", width)
print("Channels:", channels)
```

```python
# Extract RGB channels

red_channel = img[:, :, 0]
green_channel = img[:, :, 1]
blue_channel = img[:, :, 2]

print("Red Channel Shape:", red_channel.shape)
print("Green Channel Shape:", green_channel.shape)
print("Blue Channel Shape:", blue_channel.shape)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(red_channel, cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(green_channel, cmap="gray")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blue_channel, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Step 7: Convert RGB to Grayscale

gray_img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

plt.figure(figsize=(6, 6))
plt.imshow(gray_img, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
# Step 8: Image Thresholding

_, threshold_img = cv2.threshold(
    gray_img, 127, 255, cv2.THRESH_BINARY
)

plt.figure(figsize=(6, 6))
plt.imshow(threshold_img, cmap="gray")
plt.title("Threshold Image")
plt.axis("off")
plt.show()
```

```python
# Step 9: Image Blurring

blur_img = cv2.GaussianBlur(img, (15, 15), 0)

plt.figure(figsize=(6, 6))
plt.imshow(blur_img)
plt.title("Blurred Image")
plt.axis("off")
plt.show()
```

```python
# Step 10: Edge Detection

edges = cv2.Canny(gray_img, 100, 200)

plt.figure(figsize=(6, 6))
plt.imshow(edges, cmap="gray")
plt.title("Edge Detection")
plt.axis("off")
plt.show()
```

```python
# Step 11: Resize Image

resized_img = cv2.resize(img, (300, 300))

print("Original Image Shape:", img.shape)
print("Resized Image Shape:", resized_img.shape)

plt.figure(figsize=(6, 6))
plt.imshow(resized_img)
plt.title("Resized Image")
plt.axis("off")
plt.show()
```

```python
# Step 12: Rotate Image

rotated_img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

plt.figure(figsize=(6, 6))
plt.imshow(rotated_img)
plt.title("Rotated Image")
plt.axis("off")
plt.show()
```

```python
# Step 13: Flip Image

flipped_img = cv2.flip(img, 1)

plt.figure(figsize=(6, 6))
plt.imshow(flipped_img)
plt.title("Flipped Image")
plt.axis("off")
plt.show()
```

```python
# Step 14: Save Processed Images

cv2.imwrite("grayscale.jpg", gray_img)
cv2.imwrite("blurred.jpg", cv2.cvtColor(blur_img, cv2.COLOR_RGB2BGR))
cv2.imwrite("edges.jpg", edges)
cv2.imwrite("resized.jpg", cv2.cvtColor(resized_img, cv2.COLOR_RGB2BGR))
cv2.imwrite("rotated.jpg", cv2.cvtColor(rotated_img, cv2.COLOR_RGB2BGR))
cv2.imwrite("flipped.jpg", cv2.cvtColor(flipped_img, cv2.COLOR_RGB2BGR))

print("All processed images saved successfully!")
```
